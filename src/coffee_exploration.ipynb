## Part 5 – Initial Insights


Among the three datasets, the Coffee_production dataset appears to contain the most complete and consistent information. It includes 30 years of uninterrupted production data for 55 countries, with no missing values, no duplicates, and clean formatting. The structure is highly regular, making it ideal for time‑series analysis, trend detection, and comparative studies across nations.

The dataset most likely to require extensive cleaning is either Coffee_import or Coffee_export, depending on their formatting and completeness. These datasets often contain irregular country names, missing trade values, or inconsistent year ranges. Together, the three datasets can help a global coffee corporation answer strategic questions such as: Which countries are increasing or decreasing production over time? How do import and export patterns shift across regions? Which markets show growth potential, and where might supply risks emerge? These insights support decisions in sourcing, logistics, marketing, and long‑term planning.
